```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the CSV file
df = pd.read_csv("Detailed_Polling_Data.csv")  # Replace with your actual file path

# 2. Select the core parties for clustering
party_cols = [
    'Dravida Munnetra Kazhagam', 
    'All India Anna Dravida Munnetra Kazhagam', 
    'Naam Tamilar Katchi', 
    'Tamilaga Vettri Kazhagam'
]

# 3. Handle missing values and convert absolute votes to shares
df[party_cols] = df[party_cols].fillna(0)

# Calculate share using the cleaned total valid votes column from your dataset
df['Core_Total_Votes'] = df[party_cols].sum(axis=1)

# Prevent division by zero for empty booths
df = df[df['Core_Total_Votes'] > 0].copy()

# Generate percentage share features
share_cols = []
for party in party_cols:
    col_name = f'{party}_Share'
    df[col_name] = df[party] / df['Core_Total_Votes']
    share_cols.append(col_name)

# 4. Prepare data for machine learning
X = df[share_cols]

# Scale features to eliminate variance bias
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 5. Automatically find the Elbow point
wcss = []
k_range = range(1, 8)
for k in k_range:
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

# Plot Elbow Curve
plt.figure(figsize=(8, 4))
plt.plot(k_range, wcss, marker='o', color='purple')
plt.title('Elbow Method to Find Optimal Voter Profile Clusters')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Inertia)')
plt.grid(True)
plt.show()
# 6. Apply K-Means CORRECTLY using fit_predict to get cluster labels (0, 1, 2, 3)
optimal_k = 4
kmeans = KMeans(n_clusters=optimal_k, init='k-means++', random_state=42, n_init=10)
df['Cluster_ID'] = kmeans.fit_predict(X_scaled)  # <-- Fixed line

# 7. Print the correct Profile Breakdown of each individual cluster
print("\n--- TRUE AVERAGE VOTE SHARE PER CLUSTER ---")
profile = df.groupby('Cluster_ID')[share_cols].mean() * 100
print(profile.round(2))

print("\n--- BOOTH COUNT PER CLUSTER ---")
print(df['Cluster_ID'].value_counts())


# 7. Print the Profile Breakdown of each cluster
print("\n--- AVERAGE VOTE SHARE PER CLUSTER ---")
profile = df.groupby('Cluster_ID')[share_cols].mean() * 100
print(profile.round(2))

print("\n--- BOOTH COUNT PER CLUSTER ---")
print(df['Cluster_ID'].value_counts())

# Save the segmented results back to a new CSV file
df.to_csv("clustered_polling_stations.csv", index=False)
print("\nSuccess! Saved segmented data to 'clustered_polling_stations.csv'")
```

```python
# 1. Filter only the booths belonging to Cluster 2 (The TVK Dominance cluster)
tvk_dominant_booths = df[df['Cluster_ID'] == 2][
    ['Serial No. Of Polling Station', 'Locality', 'Building in Which it will be Located', 'Polling Area', 'Tamilaga Vettri Kazhagam_Share']
]

# 2. Sort them so the absolute highest TVK-performing booths appear first
tvk_dominant_booths = tvk_dominant_booths.sort_values(by='Tamilaga Vettri Kazhagam_Share', ascending=False)

# 3. Export to a clean CSV file
tvk_dominant_booths.to_csv("TVK_Priority_Booths_List.csv", index=False)

print(f"Success! Saved {len(tvk_dominant_booths)} booths to 'TVK_Priority_Booths_List.csv'.")
print(tvk_dominant_booths.head(5)) 


# 4. Run the Independent candidate analysis using the correct 'df' and 'Cluster_ID'
independent_analysis = df.groupby('Cluster_ID')[['Total_Independent_Votes', 'Margin_Percentage']].mean()

print("\n--- INDEPENDENT CANDIDATE IMPACT ANALYSIS ---")
print(independent_analysis.round(2))
```

```python
# Extract and export Cluster 1 (The DMK Strongholds)
dmk_strongholds = df[df['Cluster_ID'] == 1][
    ['Serial No. Of Polling Station', 'Locality', 'Building in Which it will be Located', 'Polling Area', 'Dravida Munnetra Kazhagam_Share']
].sort_values(by='Dravida Munnetra Kazhagam_Share', ascending=False)

dmk_strongholds.to_csv("DMK_Strongholds_Booths_List.csv", index=False)


# Extract and export Cluster 3 (The AIADMK Fortresses)
aiadmk_fortresses = df[df['Cluster_ID'] == 3][
    ['Serial No. Of Polling Station', 'Locality', 'Building in Which it will be Located', 'Polling Area', 'All India Anna Dravida Munnetra Kazhagam_Share']
].sort_values(by='All India Anna Dravida Munnetra Kazhagam_Share', ascending=False)

aiadmk_fortresses.to_csv("AIADMK_Fortresses_Booths_List.csv", index=False)


# Extract and export Cluster 0 (The Fragmented/High-Risk Swing Booths)
swing_booths = df[df['Cluster_ID'] == 0][
    ['Serial No. Of Polling Station', 'Locality', 'Building in Which it will be Located', 'Polling Area', 'Margin_Percentage']
].sort_values(by='Margin_Percentage', ascending=True) # Sorted by lowest margin (most vulnerable booths)

swing_booths.to_csv("High_Risk_Swing_Booths_List.csv", index=False)

print("All campaign lists exported successfully!")
print(f"- DMK Strongholds: {len(dmk_strongholds)} booths saved.")
print(f"- AIADMK Fortresses: {len(aiadmk_fortresses)} booths saved.")
print(f"- High-Risk Swing Zone: {len(swing_booths)} booths saved.")
```

```python
# Filter down strictly to the 62 Swing Booths (Cluster 0)
cluster_0_data = df[df['Cluster_ID'] == 0].copy()

# Feature Engineering: Identify high-risk "Spoiler Booths"
# A booth is a 'High_Spoiler_Risk' if independent votes exceed the margin percentage threshold
cluster_0_data['Independent_vs_Margin_Ratio'] = cluster_0_data['Total_Independent_Votes'] / (cluster_0_data['Margin_Percentage'] + 0.1)

# Sort the booths: Highest ratio means independents have the highest power to flip the winner
vulnerable_swing_booths = cluster_0_data[
    ['Serial No. Of Polling Station', 'Locality', 'Polling Area', 'Total_Independent_Votes', 'Margin_Percentage', 'Independent_vs_Margin_Ratio']
].sort_values(by='Independent_vs_Margin_Ratio', ascending=False)

# Export this specific high-priority list
vulnerable_swing_booths.to_csv("Cluster_0_Vulnerable_Swing_Booths.csv", index=False)

print("--- TOP 5 MOST VULNERABLE SWING BOOTHS IN CLUSTER 0 ---")
print(vulnerable_swing_booths.head(5))
```
